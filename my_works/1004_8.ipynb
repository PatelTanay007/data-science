```python
class employee1:
    def __init__(self):
        self.__salary=10000
    def show_salary(self):
        print(self.__salary)
    def set_salary(self,a):
        self.__salary=a
        print(f"new salary value is {self.__salary}")
a=employee1()

a.show_salary()
a.set_salary(15000)
a.show_salary()
```
