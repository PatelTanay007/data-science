```python


#enumerate() saperates indexing and elementd

students=["a","b","c","d"]

for i in range(len(students)):
    print(i)

for i,j in enumerate(students):

    print(i,j)

for i,j in enumerate(students,start=1001):

    print(i,j)

```

```python
#zip

students=["a","b","c","d"]
marks=[11,12,13,11,10]

for i in range(len(students)):
    print(students[i],marks[i])

print()
print()

for a,b in zip(students,marks):
    print(a,b)
```

```python
# 0 ,1,1,2,3,5,8,13....

def fib_ser(n):
    a,b = 0,1
    for i in range(n):
        yield a
        a,b = b,a+b
```

```python
a = fib_ser(15)

for i in a:
    print(i)
```
