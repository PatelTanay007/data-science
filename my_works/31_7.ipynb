```python
#multiple inheritance

class First:
    def workin(self):
        print("workin functiom")
    pass

class Second:
    def notworkin(self):
        print("this is notworkin function")
    pass

class Third(First,Second):
    def third(self):
        print("this is new")
    pass

th=Third()

th.workin()
th.notworkin()
th.third()
```

```python
#multi level inheritance

class Four:
    def four(self):
        print("fourth functiom")
    pass

class Five(Four):
    def five(self):
        print("fifth function")
    pass

class Six(Five):
    def six(self):
        print("sixth function")
    pass

s=Six()

s.four()
s.five()
s.six()
```

```python
#hyrarchicle inheritance


class Four:
    def four(self):
        print("fourth functiom")
    pass

class Five(Four):
    def five(self):
        print("fifth function")
    pass

class Six(Five):
    def six(self):
        print("sixth function")
    pass

s=Six()

s.four()
s.five()
s.six()
```
