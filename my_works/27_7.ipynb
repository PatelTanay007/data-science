```python
try:
    a=int(input("enter the first number"))
    b=int(input("enter the second number"))
    result=a/b
except Exception as e:
    print("exception" ,e)
else:
    print(f"your ans is {result}")
```

```python
try:
    a=int(input("enter the first number"))
    b=int(input("enter the second number"))
    result=a/b
except ValueError:
    print("please enter the integer value")
except ZeroDivisionError:
    print("can not devide with zero")
else:
    print(f"your ans is {result}")
finally:
    print("your program has ended")
```
