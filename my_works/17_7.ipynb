```python
num=int(input("enter the number"))
fact=1
i=1
while i<num+1:
    fact=fact*i
    i+=1

print(f"factorial of given. umber is {fact}")
```

```python
a=input("enter the string")
vowels=0
l=["a","e","i","o","u"]

for i in a:
    for j in l:
        if i == j:
            vowels+=1
print(f"total number of vowels in this string are {vowels}")
```
