```python
class Tops:
    def wlc_msg(a):
        print("hello everyone")
    pass

c=Tops()

c.wlc_msg()
```

```python
class Student:
    def __init__(self) :
        self.name = "nirav"
    def wlc_msg(self):
        print(f"hello {self.name}")
    pass

d=Student()

d.wlc_msg()
```

```python
class Student1:
    def __init__(self,name,age) :
        self.name = name
        self.age = age
    
    def wlc_msg(self):
        print(f"hello {self.name} and your age is {self.age}")
    pass

o=Student1("tanay",21)

o.wlc_msg()
```

```python
class Mota:
    def __init__(self , name , email_id , phone_nu , age):
        self.name = name
        self.email_id = email_id
        self.phone_nu = phone_nu
        self.age = age

    def return_detailes(self):
        return self.name, self.email_id , self.phone_nu , self.age


nirav = Mota("Nirav","nerav<3A@gmail.com",1234567890,67)

nirav.return_detailes()
```
