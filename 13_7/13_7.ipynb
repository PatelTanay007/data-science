```python
item_price=90
delivery_fee=10
coupon_discount=20
# Calculate the total price
total_price=item_price + delivery_fee - coupon_discount

print("value of the item =", item_price)
print("delivery fee =", delivery_fee)
print("coupon discount =", coupon_discount)
print("Total price:", total_price)
```

task 2

```python
'''check if the item price is greater than 1000
if yes then remove the delivery fee from the total price and print "eligible for free delivery"'''
if item_price > 1000:
    print("eligible for free delivery")
    total_price=total_price - delivery_fee
else:
    print("delivery charge will be applied")

print("final price:", total_price)
```
