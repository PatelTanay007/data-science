```python
import sys
import subprocess

# Create a virtual environment
venv_folder = "venv_folder"
subprocess.run([sys.executable, "-m", "venv", venv_folder])

# Find pip inside the virtual environment
if os.name == "nt":
    pip_path = os.path.join(venv_folder, "Scripts", "pip.exe")
    python_path = os.path.join(venv_folder, "Scripts", "python.exe")
else:
    pip_path = os.path.join(venv_folder, "bin", "pip")
    python_path = os.path.join(venv_folder, "bin", "python")

# Install requests package
subprocess.run([pip_path, "install", "requests"])

# Import requests from the virtual environment
result = subprocess.run(
    [python_path, "-c", "import requests; print(requests.__version__)"],
    capture_output=True,
    text=True
)

print("Virtual environment created and requests installed.")
print("Requests version:", result.stdout.strip())
```
