```python
steps = [7500, 8200, 9800, 10500, 12000, 9500, 11000]
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

i = 0

while i < len(steps):
    if steps[i] > 10000:
        print("First day you crossed 10,000 steps:", days[i])
        break
    i += 1
```

```python
from turtle import done


cart_price=[200,0,499,500,649,200,300,700,249]
total=0

done=[]

for i in cart_price:
    if i==0:
        continue
    
    total=total+i
    done.append(i)
    if total>=2000:
        break

print(f"final payablevammount is rs{total}")

remainin_items_price=cart_price

for i in done :
    remainin_items_price.remove(i)

total_remaining=sum(remainin_items_price)
print(f"the remaining items are {remainin_items_price} and its total ammount is{total_remaining}")
```
