```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


# Create a Product object
product = Product("Backpack", 1000)

# Print discounted price
print("Product:", product.name)
print("Original Price: ₹", product.price)
print("Discounted Price: ₹", product.get_discounted_price())
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


class Electronics(Product):
    def get_discounted_price(self):
        return self.price * 0.80


# Create an Electronics object
laptop = Electronics("Laptop", 50000)

print("Product:", laptop.name)
print("Original Price: ₹", laptop.price)
print("Discounted Price: ₹", laptop.get_discounted_price())
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        return self.price * 0.90


class Electronics(Product):
    def get_discounted_price(self):
        return self.price * 0.80


def show_final_price(item):
    print("Product:", item.name)
    print("Final Price: ₹", item.get_discounted_price())
    print()


# Create objects
product = Product("Backpack", 1000)
electronics = Electronics("Laptop", 50000)

# Demonstrate polymorphism
show_final_price(product)
show_final_price(electronics)
```

```python
class Ticket:
    def __init__(self, price):
        self.price = price

    def get_final_price(self):
        return self.price


class PremiumTicket(Ticket):
    def get_final_price(self):
        return super().get_final_price() + 50


# Create ticket objects
regular_ticket = Ticket(200)
premium_ticket = PremiumTicket(200)

# Print final prices
print("Regular Ticket Final Price: ₹", regular_ticket.get_final_price())
print("Premium Ticket Final Price: ₹", premium_ticket.get_final_price())
```
