```python
class Song:
    def __init__(self, title, artist, duration):
        self.title = title
        self.artist = artist
        self.duration = duration


# Create a Song object
favorite_song = Song("Blinding Lights", "The Weeknd", 200)

# Print song details
print("Song Details:")
print("Title:", favorite_song.title)
print("Artist:", favorite_song.artist)
print("Duration:", favorite_song.duration, "seconds")
```

```python
class Song:
    def __init__(self, title, artist, duration):
        self.title = title
        self.artist = artist
        self.duration = duration

    def play_preview(self):
        print(f"Playing 30-second preview of {self.title} by {self.artist}")


# Create a Song object
favorite_song = Song("Blinding Lights", "The Weeknd", 200)

# Play the preview
favorite_song.play_preview()
```

```python
class FoodOrder:
    def __init__(self, restaurant_name, items, total_price):
        self.restaurant_name = restaurant_name
        self.items = items
        self.total_price = total_price


# Create a FoodOrder object
last_order = FoodOrder(
    "Domino's Pizza",
    ["Margherita Pizza", "Garlic Bread"],
    450
)

# Print order details
print("Food Order Details:")
print("Restaurant:", last_order.restaurant_name)
print("Items:", last_order.items)
print("Total Price: ₹", last_order.total_price)
```

```python
class FoodOrder:
    def __init__(self, restaurant_name, items, total_price):
        self.restaurant_name = restaurant_name
        self.items = items
        self.total_price = total_price

    def add_item(self, item_name, item_price):
        self.items.append(item_name)
        self.total_price += item_price


# Create a FoodOrder object
last_order = FoodOrder(
    "Domino's Pizza",
    ["Margherita Pizza"],
    300
)

# Add two items
last_order.add_item("Garlic Bread", 100)
last_order.add_item("Coke", 50)

# Print updated order
print("Updated Food Order:")
print("Restaurant:", last_order.restaurant_name)
print("Items:", last_order.items)
print("Updated Total Price: ₹", last_order.total_price)
```

```python
class Song:
    def __init__(self, title, artist, duration=0):
        self.title = title
        self.artist = artist
        self.duration = duration


# Create a song without providing duration
song1 = Song("Starboy", "The Weeknd")

# Create a song with duration
song2 = Song("Blinding Lights", "The Weeknd", 200)

# Print details
print("Song 1:")
print("Title:", song1.title)
print("Artist:", song1.artist)
print("Duration:", song1.duration, "seconds")

print("\nSong 2:")
print("Title:", song2.title)
print("Artist:", song2.artist)
print("Duration:", song2.duration, "seconds")
```
