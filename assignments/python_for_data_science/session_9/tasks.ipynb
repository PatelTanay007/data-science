# task 1

```python
def final_price(price, discount_rate):
    discount=price*(discount_rate/100)
    final_price=price-discount
    return final_price

item_price=1000
discount_per=20

payable_ammount=final_price(item_price,discount_per)

print (f"final payable ammount is{payable_ammount}")
```

```python
def format_price(price, currency="INR"):
    price=str(price)
    if currency=="INR":
        s="₹"+ price
    elif currency=="USD":
        s="$"+ price
    else:
        s="please enter valid currency"
    return(s)

p=int(input("enter the price"))

c=input("enter the currency")

ot=format_price(p,c)

print(ot)
    
```
