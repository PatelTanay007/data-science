```python
def print_playlist_songs(songs):
    if not songs:
        return
    print(songs[0])
    print_playlist_songs(songs[1:])

playlist = [
    "Blinding Lights",
    "Shape of You",
    "Believer",
    "Levitating",
    "Perfect"
]

print("Playlist Songs:")
print_playlist_songs(playlist)
```

# task 2

```python
def count_unread_messages(messages):
    total = messages.get("count", 0)

    for subgroup in messages.get("subgroups", []):
        total += count_unread_messages(subgroup)

    return total

whatsapp_groups = {
    "name": "Family",
    "count": 5,
    "subgroups": [
        {
            "name": "Cousins",
            "count": 3,
            "subgroups": []
        },
        {
            "name": "Friends",
            "count": 7,
            "subgroups": [
                {
                    "name": "College",
                    "count": 4,
                    "subgroups": []
                },
                {
                    "name": "Office",
                    "count": 2,
                    "subgroups": []
                }
            ]
        }
    ]
}

print("Total Unread Messages:", count_unread_messages(whatsapp_groups))
```

# task 3

```python
x = "global"  # Global variable

def outer():
    x = "outer"  # Local variable in outer()

    def inner():
        nonlocal x  # Refers to x from outer()
        x = "inner"  # Changes outer()'s x

    inner()
    print("Inside outer:", x)

outer()
print("Outside:", x)

# Answer:
# Global Variable:
# - x = "global"

# Local Variable:
# - x = "outer" (local to outer())

# Explanation:
# - inner() uses 'nonlocal x', so it modifies outer()'s x.
# - outer()'s x changes from "outer" to "inner".
# - The global x remains unchanged.

# Output:
# Inside outer: inner
# Outside: global
```

```python
def format_number_short(n):
    if n < 1000:
        return str(n)
    elif n < 1000000:
        return f"{n / 1000:.1f}K"
    elif n < 1000000000:
        return f"{n / 1000000:.1f}M"
    else:
        return f"{n / 1000000000:.1f}B"

numbers = [500, 1500, 25000, 1200000, 2500000000]

print("Formatted Numbers:")
for num in numbers:
    print(f"{num} -> {format_number_short(num)}")
```
