# tsak 3

```python
import csv

with open("data.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    print("Season\t     Date\t\tTeam 1\t\t\tTeam 2\t\t\t\tWinner")
    print("-" * 110)

    for row in reader:
        print(
            row["season"], "\t",
            row["date"], "\t",
            row["team1"], "\t",
            row["team2"], "\t",
            row["winner"]
        )
```

```python
import json

with open("Film.json", "r") as file:
    movies = json.load(file)

for movie in movies:
    print("Title:", movie["Title"])
    print("IMDb Rating:", movie["imdbRating"])
    print("-" * 30)
```

# tsak 5

```python
from pathlib import Path
import json

file_path = Path("my_fav_apps.json")

if not file_path.exists():
    apps = [
        {"name": "Instagram", "category": "Social Media"},
        {"name": "Zomato", "category": "Food Delivery"},
        {"name": "Paytm", "category": "Payments"}
    ]

    with open(file_path, "w") as file:
        json.dump(apps, file, indent=4)

    print("my_fav_apps.json created successfully.")
else:
    print("my_fav_apps.json already exists.")
```
