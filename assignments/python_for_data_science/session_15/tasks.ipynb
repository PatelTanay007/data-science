```python
def get_song_duration_per_minute(total_duration_minutes, number_of_songs):
    try:
        average_duration = total_duration_minutes / number_of_songs
        print(f"Average song duration: {average_duration:.2f} minutes")
        return average_duration

    except ZeroDivisionError:
        print("The playlist contains zero songs.")
        return None

    finally:
        print("Duration calculation completed.")


get_song_duration_per_minute(120, 24)

get_song_duration_per_minute(120, 0)
```

# task 2

```python
try:
    total_cart_amount = float(input("Enter total cart amount (₹): "))
    item_count = int(input("Enter number of items: "))

    price_per_item = total_cart_amount / item_count

    print(f"Price per item: ₹{price_per_item:.2f}")

except ZeroDivisionError:
    print("Oops! Item count cannot be zero. Please add at least one item to the cart.")

except ValueError:
    print("Invalid input! Please enter numeric values only.")
```

# task 3

```python
class customee(Exception):
    pass

try:
    total_spend = float(input("Enter total spend (₹): "))
    offers_applied = int(input("Enter number of offers applied: "))

    if offers_applied == 0:
        raise customee("No cashback offers were applied.")

    average_cashback = total_spend / offers_applied

    print(f"Average cashback per offer: ₹{average_cashback:.2f}")

except customee as e:
    print(f"Error: {e} Please apply at least one offer.")

except ValueError:
    print("Invalid input! Please enter numeric values only.")
```

```python
def calculate_average_rating(total_rating, num_reviews):
    return total_rating / num_reviews

try:
    result = calculate_average_rating(500, 0)
    print("Average Rating:", result)

except ZeroDivisionError:
    print("Error: Number of reviews cannot be zero.")

finally:
    print("Thank you for using the calculator")
```

```python
def safe_divide_for_zomato(bill_amount, number_of_people):
    try:
        result = bill_amount / number_of_people

    except ZeroDivisionError:
        print("Error: Cannot split the bill among 0 people.")

    else:
        print(f"Each person should pay: ₹{result:.2f}")

    finally:
        print("Split calculation done")


# Example 1
safe_divide_for_zomato(1200, 4)

print()

# Example 2
safe_divide_for_zomato(1200, 0)
```
